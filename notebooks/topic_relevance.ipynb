# Topic relevance diagnostic

The local Python diagnostic was executed on 18 November 2026. The accompanying source and authored dialogue set are included in this project. This notebook presents the saved outputs. Natural dialogue and independent annotation remain pending.

```python
from pathlib import Path
import json
import pandas as pd
from IPython.display import display

root = Path.cwd() if Path("results/topic_policy.json").is_file() else Path.cwd().parent
results = root / "results"
policy = json.loads((results / "topic_policy.json").read_text())
metrics = json.loads((results / "topic_test_metrics.json").read_text())
latency = json.loads((results / "topic_latency.json").read_text())
display(pd.Series(policy, name="Selected policy").to_frame())
display(pd.Series(metrics, name="Reserved-topic result").to_frame())
display(pd.Series(latency, name="Warm CPU latency").to_frame())
```

## Limits found in the reserved examples
